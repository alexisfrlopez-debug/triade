## 2 · LE KIT AU COFFRE — monter le Drive, tirer le kit poste v100 TEL QUEL, recouper sha256, extraire

Le coffre vit au Drive monté : `Triade/livrables` (la voie directe), le repli
à la marche bornée du `MyDrive` au nom exact, le repli final à l'API Drive à
l'ID épinglé. **L'empreinte sha256 est épinglée** : le zip divergent ne court
pas (KIT-DIVERGENT, fail-closed).

**LE TERRAIN.6 CORRIGÉ (v102)** : la cellule PARLE à chaque pas — la fenêtre
d'AUTORISATION est annoncée AVANT de survenir (le bloqueur de pop-ups nommé,
le ~15-30 s du montage nommés), un montage DÉJÀ FAIT n'est PAS redemandé, la
marche du MyDrive vit à LA BOURSE de 45 s au COMPTE À VOIX HAUTE (dossiers
vus, secondes), et le repli API tire le kit MORCEAU PAR MORCEAU (le pourcentage
imprimé) — le silence n'est plus jamais pris pour un blocage.

```python
# CELLULE 2 — LE KIT AU COFFRE : monter le Drive, tirer le kit poste v100 TEL QUEL,
# recouper sha256, extraire. Le kit Colab ne porte AUCUN moteur : il court le moteur
# du kit poste v100 byte-byte (l'ID et l'empreinte ÉPINGLÉS — KIT-DIVERGENT le refus).
# LE TERRAIN.6 DU 27/11, CORRIGÉ (v102) : le v101 SE TAISAIT aux DEUX seuils
# interactifs (la fenêtre d'autorisation du montage, celle du repli API) et durant
# LA MARCHE du MyDrive (des minutes muettes au Drive touffu) — le maître crut au
# blocage (« ça bloque à la cellule 2 »). Le v102 ANNONCE chaque seuil AVANT qu'il
# ne survienne, NE REDEMANDE PAS un montage déjà fait (ismount), BORNE la marche à
# une bourse de secondes AU COMPTE À VOIX HAUTE, et fait PARLER le repli API par
# morceaux — la cellule muette est morte.
import hashlib
import os
import pathlib
import shutil
import time
import zipfile

NOM_KIT = "TRIADE_kit_campagne_poste_v100.zip"
SHA256_KIT = "a8d9b96c17ccdc4d1ca72fa1964705fe09520096de8ef17ae3eb88366f53642b"
ID_KIT_DRIVE = "1TT1zLRf1movZ69_cTzGhMLEcAV1Hi7CD"    # le /livrables du coffre (l'Entrée 443)
CHEMIN_DIRECT = pathlib.Path("/content/drive/MyDrive/Triade/livrables") / NOM_KIT
DOS_KIT = pathlib.Path("/content/kit")
BOURSE_MARCHE = 45.0    # secondes — la marche du MyDrive n'ira pas au-delà


def sha256f(p):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        for morceau in iter(lambda: f.read(1 << 20), b""):
            h.update(morceau)
    return h.hexdigest()


def trouve(nom):
    """La voie directe, puis la marche BORNÉE du MyDrive au nom exact du zip —
    la bourse tenue, le compte à voix haute : la marche muette est morte."""
    if CHEMIN_DIRECT.exists():
        return CHEMIN_DIRECT
    racine = pathlib.Path("/content/drive/MyDrive")
    vu, depart = 0, time.time()
    for base, dossiers, fichiers in os.walk(racine):
        vu += 1
        if time.time() - depart > BOURSE_MARCHE:
            print(f"[coffre] la marche du MyDrive BORNÉE à {int(time.time() - depart)} s "
                  f"({vu} dossiers vus — {nom} n'y vivait pas) : le repli API prend le relais")
            return None
        profondeur = len(pathlib.Path(base).relative_to(racine).parts)
        if profondeur >= 6:
            dossiers[:] = []
        dossiers[:] = [d for d in dossiers
                       if not d.startswith(".") or d == ".shortcut-targets-by-id"]
        if vu % 100 == 0:
            print(f"[coffre] la marche du MyDrive court : {vu} dossiers vus "
                  f"({int(time.time() - depart)} s de bourse)")
        if nom in fichiers:
            print(f"[coffre] le zip trouvé à la marche : {base}/{nom} "
                  f"({vu} dossiers vus, {int(time.time() - depart)} s)")
            return pathlib.Path(base) / nom
    return None


MONTURE = "/content/drive"
if os.path.ismount(MONTURE):
    print("[coffre] le Drive est DÉJÀ MONTÉ — l'autorisation n'est PAS redemandée")
else:
    print("[coffre] LE SEUIL INTERACTIF : une FENÊTRE D'AUTORISATION s'ouvre sous cette cellule")
    print("[coffre] (« Se connecter à Google Drive ») — clique AUTORISER ; si la fenêtre")
    print("[coffre] ne paraît pas, lève le bloqueur de pop-ups du navigateur puis relance ;")
    print("[coffre] le montage prend ~15-30 s après le clic — la cellule reparle ensuite.")
    from google.colab import drive
    drive.mount(MONTURE)
print("[coffre] le Drive monté — la quête du kit court")

zip_kit = trouve(NOM_KIT)
if zip_kit is None:
    # le repli final : l'API Drive à l'ID épinglé (le compte monté a la main sur le coffre)
    print("[coffre] le zip est introuvable au MyDrive monté — LE REPLI API à l'ID épinglé court")
    print("[coffre] LE SEUIL INTERACTIF : une SECONDE autorisation peut être demandée à la")
    print("[coffre] même fenêtre — clique AUTORISER, la cellule reprend ensuite.")
    import google.auth
    from google.colab import auth
    from googleapiclient.discovery import build
    from googleapiclient.http import MediaIoBaseDownload
    auth.authenticate_user()
    creds, _ = google.auth.default()
    drive_svc = build("drive", "v3", credentials=creds)
    zip_kit = pathlib.Path("/content") / NOM_KIT
    with open(zip_kit, "wb") as f:
        telecharge = MediaIoBaseDownload(f, drive_svc.files().get_media(fileId=ID_KIT_DRIVE))
        fait = False
        while not fait:
            progres, fait = telecharge.next_chunk()
            print(f"[coffre] le repli API tire le kit : {int(progres.progress() * 100)} %")
    print(f"[coffre] le repli API a tiré le zip à l'ID épinglé {ID_KIT_DRIVE} — VERT")

if zip_kit is None:
    raise SystemExit("KIT-ABSENT : le zip du kit est introuvable au Drive monté — "
                     "le LISEZMOI porte la voie.")

empreinte = sha256f(zip_kit)
print(f"[coffre] le kit trouvé : {zip_kit}")
print(f"[coffre] sha256 : {empreinte}")
if empreinte != SHA256_KIT:
    raise SystemExit(f"KIT-DIVERGENT : {empreinte} != {SHA256_KIT} (l'empreinte épinglée) — "
                     "le kit NE COURT PAS (fail-closed).")
print("[coffre] l'empreinte recoupée VERT — le kit poste v100 TEL QUEL, byte-byte.")

if DOS_KIT.exists():
    shutil.rmtree(DOS_KIT)
with zipfile.ZipFile(zip_kit) as z:
    z.extractall(DOS_KIT)
pieces = [p for p in DOS_KIT.rglob("*") if p.is_file()]
print(f"[coffre] le kit extrait à {DOS_KIT} : {len(pieces)} pièces")

COFFRE_TEST = pathlib.Path("/content/drive/MyDrive/Triade/test")
print(f"[coffre] le dossier /test du coffre (le retour de la cellule 8) : "
      f"{'VERT — ' + str(COFFRE_TEST) if COFFRE_TEST.exists() else 'ABSENT — le déposer à la main avant le retour'}")
```

## 5 · LE SERVING MODE-B — llama.cpp au runtime + le GGUF du modèle choisi

La commande consignée au LISEZMOI-CAMPAGNE-POSTE § 1 (`llama-server -m <gguf>
--port 8080 -c 8192 -ngl 99`) reconduite au poste Colab. Le GGUF du minter
**bartowski** (Hugging Face) ; la qualité au choix du poste : `QUANT` dans la
cellule (Q4_K_M 4,68 Go · Q6_K 6,25 Go · Q8_0 8,10 Go). La release prébattue
de llama.cpp d'abord, la source compilée en repli nommé — **LE TERRAIN.5 DU
27/11 CORRIGÉ** : l'hygiène du port d'abord (les orphelins tués, le port LIBRE
choisi parmi les candidats 8080/8081/8090/18080/18081), le triage juste
(« couldn't bind » = le port, PAS le binaire : le prébâti réessayé, la source
ne compile PAS pour un port pris), le repli qui PARLE (la compilation ligne
à ligne au journal du poste), l'endpoint PORTÉ à la GUI par le fichier
`/content/endpoint_serving.txt`.

```python
# CELLULE 5 — LE SERVING MODE-B : llama.cpp au runtime + le GGUF du modèle choisi
# La commande consignée au LISEZMOI-CAMPAGNE-POSTE § 1 (llama-server -m <gguf>
# --port 8080 -c 8192 [-ngl 99]) reconduite au poste Colab. LE TERRAIN.5 DU
# 27/11 NOMMÉ : « couldn't bind … 8080 » = LE PORT OCCUPÉ, PAS le binaire —
# le v100 accusa le prébâti et lança une compilation muette ; le v101 TRIE.
QUANT = "Q6_K"          # Q4_K_M (4,68 Go) · Q6_K (6,25 Go) · Q8_0 (8,10 Go) — la qualité au choix du poste
REPO_GGUF = "bartowski/Qwen2.5-Coder-7B-Instruct-GGUF"
PORTS = [8080, 8081, 8090, 18080, 18081]   # les candidats : 8080 peut être occupé au poste (le terrain.5)

import json
import os
import pathlib
import socket
import subprocess
import tarfile
import time
import urllib.request
import zipfile


def gpu_presente():
    try:
        import torch
        return torch.cuda.is_available()
    except Exception:
        return False


GPU = gpu_presente()
DOS_LLAMA = pathlib.Path("/content/llama")
CHEMIN_ENDPOINT = pathlib.Path("/content/endpoint_serving.txt")


def port_libre(port):
    """La sonde au socket même — SO_REUSEADDR porté : la même tolérance que
    llama-server au bind (la sonde n'est jamais plus sévère que le serveur)."""
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)
    try:
        s.bind(("127.0.0.1", port))
        return True
    except OSError:
        return False
    finally:
        s.close()


def choisi_le_port(apres=None):
    """Le premier port LIBRE des candidats — après celui qui vient d'échouer
    le cas échéant (le réessai ne reprend JAMAIS le port qui a refusé)."""
    for p in PORTS:
        if apres is not None and p <= apres:
            continue
        if port_libre(p):
            return p
    return None


def menage_des_orphelins():
    """L'hygiène du terrain.5 : un llama-server orphelin d'un run précédent du
    MÊME runtime squatte le port — LUI SEUL est tué (jamais un tiers du poste)."""
    r = subprocess.run(["pkill", "-9", "-f", "llama-server"], capture_output=True)
    tue = r.returncode == 0
    if tue:
        time.sleep(2)      # le relâchement du port par le noyau
    return tue


def asset_de_la_release():
    """La release PRÉBATTIE du jour : les tags b portent les binaires (les tags v
    n'en portent pas — vérifié au bac : 38 assets au b12810). Le choix :
    GPU → le cudart (le runtime CUDA embarqué, tolérant au driver du poste) ;
    CPU → le ubuntu-x64 nu. Retour None si l'API ou le choix manque — la source
    prend le relais (le repli NOMMÉ)."""
    try:
        with urllib.request.urlopen(
                "https://api.github.com/repos/ggml-org/llama.cpp/releases?per_page=12",
                timeout=30) as r:
            releases = json.loads(r.read().decode("utf-8"))
    except Exception as e:
        print(f"[serving] l'API des releases a manqué ({e.__class__.__name__}) — la SOURCE prend le relais.")
        return None
    for rel in releases:
        nu, cudarts = None, []
        for a in rel["assets"]:
            n = a["name"].lower()
            if "ubuntu" not in n or "x64" not in n or not n.endswith((".zip", ".tar.gz")):
                continue
            exclus = ("arm64", "s390x", "vulkan", "rocm", "sycl", "openvino", "opencl")
            if any(x in n for x in exclus):
                continue
            if n.startswith("cudart-") and "cuda" in n:
                cudarts.append(a)
            if "cuda" not in n and not n.startswith("cudart-"):
                nu = a
        if GPU and cudarts:
            # le cudart CUDA-12 d'abord (plus tolérant au driver du poste que le 13)
            choix = next((a for a in cudarts if "cuda-12" in a["name"].lower()), cudarts[0])
            return rel["tag_name"], choix
        if not GPU and nu is not None:
            return rel["tag_name"], nu
    return None


def extrais():
    """L'extraction du prébâti : les tags b sortent le dossier <llama-bXXXXX> à
    plats (binaires + libs .so frères — vérifié au bac)."""
    tag_asset = asset_de_la_release()
    if tag_asset is None:
        return None
    tag, asset = tag_asset
    print(f"[serving] la release prébattue {tag} : {asset['name']} "
          f"({asset['size'] / 2**20:.0f} Mio)")
    archive = pathlib.Path("/content") / asset["name"]
    if archive.exists():
        archive.unlink()
    urllib.request.urlretrieve(asset["browser_download_url"], archive)
    if asset["name"].endswith(".zip"):
        with zipfile.ZipFile(archive) as z:
            z.extractall(DOS_LLAMA)
    else:
        with tarfile.open(archive) as t:
            try:
                t.extractall(DOS_LLAMA, filter="data")
            except TypeError:
                t.extractall(DOS_LLAMA)     # le Python d'avant 3.12 ignore le filtre
    trouves = list(DOS_LLAMA.rglob("llama-server"))
    return trouves[0] if trouves else None


def compile_la_source():
    """LE REPLI QUI PARLE : le terrain.5 nomma le v100 muet (la compilation
    capturée, les minutes sans journal) — chaque ligne coule au poste à
    présent, la cible llama-server seule bâtie (le chantier raccourci)."""
    print("[serving] la compilation depuis la source — LE CHANTIER QUI PARLE "
          "(5-15 min aux vCPU du poste, le café est nommé ; chaque ligne ci-dessous).")
    commande = ("git clone --depth 1 https://github.com/ggml-org/llama.cpp /content/llama.cpp && "
                "cmake -S /content/llama.cpp -B /content/llama.cpp/build"
                + (" -DGGML_CUDA=ON" if GPU else "") + " && "
                "cmake --build /content/llama.cpp/build --config Release "
                "--target llama-server -j$(nproc)")
    debut = time.time()
    proc = subprocess.Popen(["bash", "-c", commande], stdout=subprocess.PIPE,
                            stderr=subprocess.STDOUT, text=True, bufsize=1,
                            encoding="utf-8", errors="replace")
    n_lignes = 0
    for ligne in proc.stdout:
        n_lignes += 1
        print(ligne, end="")
    code = proc.wait()
    print(f"[serving] la compilation close : code {code}, {n_lignes} lignes, "
          f"{(time.time() - debut) / 60:.1f} min.")
    if code != 0:
        raise SystemExit(f"[ROUGE] la compilation a échoué (code {code}) — le journal ci-dessus TEL QUEL.")
    return pathlib.Path("/content/llama.cpp/build/bin/llama-server")


def lance_serving(binaire, gguf, port):
    """Le lancement + la porte de santé (240 × 5 s) au PORT choisi. Les libs
    .so frères du binaire prébâti portent au LD_LIBRARY_PATH du poste."""
    env = dict(os.environ)
    env["LD_LIBRARY_PATH"] = str(binaire.parent) + (
        ":" + env["LD_LIBRARY_PATH"] if env.get("LD_LIBRARY_PATH") else "")
    args_serving = [str(binaire), "-m", str(gguf), "--host", "127.0.0.1",
                    "--port", str(port), "-c", "8192"]
    if GPU:
        args_serving += ["-ngl", "99"]
    journal = open("/content/llama-server.log", "w", encoding="utf-8")
    proc = subprocess.Popen(args_serving, stdout=journal, stderr=subprocess.STDOUT, env=env)
    for _ in range(240):
        time.sleep(5)
        if proc.poll() is not None:
            break
        try:
            with urllib.request.urlopen(f"http://127.0.0.1:{port}/health", timeout=10) as r:
                if r.status == 200:
                    journal.close()
                    return proc, True
        except Exception:
            pass
    journal.close()
    return proc, False


def lit_journal():
    """Le journal du serveur, TEL QUEL (le terrain.5 fut lu par ce biais)."""
    try:
        return open("/content/llama-server.log", encoding="utf-8", errors="replace").read()
    except OSError:
        return ""


# 1 · llama-server : la release prébattue d'abord, la source compilée en repli nommé
binaire = None
if DOS_LLAMA.exists():
    trouves = list(DOS_LLAMA.rglob("llama-server"))
    if trouves:
        binaire = trouves[0]
if binaire is None:
    try:
        binaire = extrais()
    except Exception as e:
        print(f"[serving] le prébâti a manqué ({e.__class__.__name__}) — la SOURCE prend le relais.")
libelle = str(binaire) if binaire is not None else "la SOURCE d'abord"
print(f"[serving] llama-server : {libelle}")

# 2 · le GGUF du modèle choisi (le vocabulaire vit au harnais — recoupé au chargement)
from huggingface_hub import hf_hub_download

gguf = hf_hub_download(repo_id=REPO_GGUF, filename=f"Qwen2.5-Coder-7B-Instruct-{QUANT}.gguf")
print(f"[serving] le GGUF {QUANT} : {gguf} "
      f"({pathlib.Path(gguf).stat().st_size / 2**30:.2f} Gio)")

# 3 · L'HYGIÈNE DU PORT (le terrain.5) : les orphelins d'un run précédent tués,
#     le port LIBRE choisi — AUCUN port n'est imposé au poste
if menage_des_orphelins():
    print("[serving] un llama-server orphelin d'un run précédent a été tué (l'hygiène du terrain.5).")
PORT = choisi_le_port()
if PORT is None:
    raise SystemExit(f"PORT-OCCUPÉ : aucun candidat n'est libre ({PORTS}) — "
                     "le journal du poste le dira ; AUCUN serving ne court (fail-closed).")
if PORT != 8080:
    print(f"[serving] le 8080 est occupé au poste — le port libre {PORT} pris (le terrain.5 nommé).")

# 4 · le serving + la porte de santé ; LE TRIAGE DU TERRAIN.5 : « couldn't bind »
#     = LE PORT, PAS LE BINAIRE — le prébâti réessayé au port suivant, la source
#     NE COMPILE PAS pour un port pris ; le binaire mort d'une faute qui est
#     sienne (glibc, driver) → la source prend le relais (le repli NOMMÉ)
proc_serving, sain = lance_serving(binaire, gguf, PORT) if binaire else (None, False)
if not sain and binaire is not None and "couldn't bind" in lit_journal():
    print("[serving] le port a refusé (couldn't bind) — le PRÉBÂTI EST SAIN, c'est le port : "
          "le réessai au port suivant, la source NE compile PAS (le triage du terrain.5).")
    if menage_des_orphelins():
        print("[serving] un second orphelin tué au réessai.")
    PORT = choisi_le_port(apres=PORT) or choisi_le_port()
    if PORT is None:
        raise SystemExit(f"PORT-OCCUPÉ : plus aucun candidat libre ({PORTS}) — "
                         "le journal du poste le dira.")
    print(f"[serving] le réessai au port {PORT}.")
    proc_serving, sain = lance_serving(binaire, gguf, PORT)
if not sain:
    if binaire is not None:
        print("[serving] le prébâti n'est pas sain au poste — le journal TEL QUEL :")
        print(lit_journal()[-1500:], end="")
        print("[serving] la SOURCE prend le relais (le repli NOMMÉ).")
    binaire = compile_la_source()
    if menage_des_orphelins():
        print("[serving] un orphelin tué avant le serving de la source.")
    PORT = choisi_le_port(apres=PORT) or choisi_le_port()
    if PORT is None:
        raise SystemExit(f"PORT-OCCUPÉ : plus aucun candidat libre ({PORTS}) après la compilation.")
    proc_serving, sain = lance_serving(binaire, gguf, PORT)
if not sain:
    print(lit_journal()[-2500:], end="")
    raise SystemExit("[ROUGE] le serving n'est pas sain — le journal ci-dessus TEL QUEL "
                     "(un échec nommé vaut plus qu'un vert muet).")
CHEMIN_ENDPOINT.write_text(f"http://127.0.0.1:{PORT}\n", encoding="utf-8")
print(f"[serving] http://127.0.0.1:{PORT}/health : VERT — le serving du modèle choisi est vivant.")
print(f"[serving] l'endpoint PORTÉ au fichier {CHEMIN_ENDPOINT} — la cellule 6 (LA GUI) "
      "le lit au champ « L'endpoint » (top-n 64).")
```

```python
# CELLULE 6 — LA GUI 1 CLIC : les CINQ BOUTONS du plan gelé (le patron de gui_campagne.py
# aux widgets Colab — la fenêtre ne fait QUE lancer le moteur et suivre son journal en
# direct : LE MOTEUR PORTE TOUT LE POIDS DU GESTE).
import pathlib
import subprocess
import sys
import threading

import ipywidgets as widgets
from IPython.display import display

MOTEUR = "/content/kit/moteur_campagne.py"
DOS_MOTEUR = "/content/kit"

entete = widgets.HTML(
    "<b>TRIADE — LA CAMPAGNE AU POSTE RÉEL · VOIE COLAB (kit v103)</b><br>"
    "les plans gelés PLAN-W29DECCO + PLAN-W29FAMILLE : aucune cible re-tirée, "
    "aucune graine re-tirée, aucun chiffre fabriqué<br>"
    "le modèle choisi : Qwen2.5-Coder-7B-Instruct — le décodage contraint CÔTÉ CLIENT (llguidance)")

# l'endpoint PORTÉ par la cellule 5 (le terrain.5 : le port qui a servi y vit)
endpoint_defaut = "http://127.0.0.1:8080"
_ep = pathlib.Path("/content/endpoint_serving.txt")
if _ep.exists():
    _lu = _ep.read_text(encoding="utf-8").strip()
    if _lu:
        endpoint_defaut = _lu

mode_w = widgets.RadioButtons(options=["B", "A"], value="B",
                             description="Mode",
                             tooltips=["B — llama.cpp (le bord natif /completion)",
                                       "A — transformers (le masque aux logits entiers)"])
url_w = widgets.Text(value=endpoint_defaut, description="L'endpoint")
nprobs_w = widgets.Text(value="64", description="top-n")
modele_w = widgets.Text(value="", description="Modèle (A)",
                        placeholder="le chemin du dossier du modèle (MODE-A seul)")
panneau = widgets.HBox([widgets.VBox([mode_w]), widgets.VBox([url_w, nprobs_w, modele_w])])

zone = widgets.Output(layout=widgets.Layout(border="1px solid #777", height="380px",
                                            overflow="auto"))
etat_w = widgets.HTML("<b>PRÊT — l'étape 1 d'abord (le manifeste)</b>")

LES_BOUTONS = [
    ("verifier-kit", "1 · Vérifier le kit (le manifeste)"),
    ("gate", "2 · Le gate (la sonde du décodeur)"),
    ("canari", "3 · Le canari (la violation refusée)"),
    ("campagne", "4 · LA CAMPAGNE (éditer + ajouter + la famille)"),
    ("zip", "5 · Le zip de preuves"),
]
boutons = {}
for nom, texte in LES_BOUTONS:
    b = widgets.Button(description=texte, layout=widgets.Layout(width="auto"))
    boutons[nom] = b

proc = {"p": None}


def _arguments(nom):
    a = [sys.executable, "-X", "utf8", MOTEUR]
    if nom == "verifier-kit":
        a.append("--verifier-kit")
    elif nom == "gate":
        a.append("--gate")
    elif nom == "canari":
        a += ["--canari", "--mode", mode_w.value]
    elif nom == "campagne":
        a += ["--campagne", "--mode", mode_w.value]
    elif nom == "zip":
        a.append("--zip")
    if mode_w.value == "B" and nom in ("canari", "campagne"):
        a += ["--url", url_w.value.strip(), "--nprobs", nprobs_w.value.strip()]
    if mode_w.value == "A" and nom in ("canari", "campagne"):
        a += ["--modele", modele_w.value.strip()]
    return a


def _grille(desactive):
    for b in boutons.values():
        b.disabled = desactive


def lancer(nom):
    if proc["p"] and proc["p"].poll() is None:
        with zone:
            print("[GUI] une étape court déjà — attends sa fin.")
        return
    args = _arguments(nom)
    with zone:
        print(f"\n[GUI] === L'ÉTAPE : {nom} ({' '.join(args[3:])}) ===")
    etat_w.value = f"<b>en cours — {nom}…</b>"
    _grille(True)
    proc["p"] = subprocess.Popen(args, cwd=DOS_MOTEUR, text=True,
                                  stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                                  encoding="utf-8", errors="replace", bufsize=1)

    def suivre():
        for ligne in proc["p"].stdout:
            zone.append_stdout(ligne)
        code = proc["p"].wait()
        zone.append_stdout(f"[ÉTAT] {'VERT' if code == 0 else 'ROUGE'} — {nom} (code {code})\n")
        if code == 0:
            etat_w.value = f"<b>VERT — {nom} (code {code})</b> — l'étape suivante"
        else:
            etat_w.value = f"<b>ROUGE — {nom} (code {code})</b> — le journal TEL QUEL ci-dessus"
        _grille(False)

    threading.Thread(target=suivre, daemon=True).start()


for nom, _ in LES_BOUTONS:
    boutons[nom].on_click(lambda _b, n=nom: lancer(n))

rang_boutons = widgets.HBox([boutons[n] for n, _ in LES_BOUTONS])
display(widgets.VBox([entete, panneau, rang_boutons, zone, etat_w]))
```

Le bouton 5 bâtit `preuves-campagne-<horodatage>.zip` au kit ; cette cellule
le dépose au dossier `Triade/test` du coffre monté, **recoupé byte-byte**
par relecture. Le registre porte le mode, les paramètres, les refus nommés —

**LE GARDE EST L'ORDRE, PAS UNE PANNE** (le terrain.7 du 27/11, nommé) :

refus nommé est le garde fail-closed du retour au coffre.

```python
# CELLULE 8 — LE RETOUR AU COFFRE : le zip de preuves déposé au /test du coffre,
# recoupé byte-byte par relecture (le geste consigné au LISEZMOI-CAMPAGNE-POSTE § 4).
import hashlib
import pathlib
import shutil

try:
    COFFRE_TEST
except NameError:
    COFFRE_TEST = pathlib.Path("/content/drive/MyDrive/Triade/test")


def sha256f(p):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        for morceau in iter(lambda: f.read(1 << 20), b""):
            h.update(morceau)
    return h.hexdigest()


zips = sorted(pathlib.Path("/content/kit").glob("preuves-campagne-*.zip"),
              key=lambda p: p.stat().st_mtime)
if not zips:
    raise SystemExit("ZIP-ABSENT : aucun preuves-campagne-*.zip au kit — "
                     "le bouton 5 de la GUI d'abord.")
source = zips[-1]
if not COFFRE_TEST.exists():
    raise SystemExit(f"COFFRE-ABSENT : le dossier {COFFRE_TEST} n'est pas monté — "
                     "le Drive est-il monté (cellule 2) ?")
destination = COFFRE_TEST / source.name
shutil.copy2(source, destination)
recoup = sha256f(destination) == sha256f(source)
empreinte = sha256f(source)
print(f"[retour] {source.name} ({source.stat().st_size} o)")
print(f"[retour] déposé à {destination}")
print(f"[retour] sha256 : {empreinte}")
if not recoup:
    raise SystemExit("RECOP-ROUGE : la pièce déposée diverge de la pièce du kit — re-déposer.")
print("[retour] recoupé byte-byte par relecture — VERT. Le labo lira les verdicts TELS QUELS.")
```

```python
# CELLULE 9 — LE PONT DU POSTE (optionnel) : l'URL du runtime VUE DE TON ORDI
# (le terrain.7 du 27/11 : le maître chercha « la web app au localhost de la
# session » — elle n'existe pas ; LA GUI 1 CLIC vit DANS LE NOTEBOOK, la
# cellule 6 ; l'URL http://127.0.0.1:PORT vit AU RUNTIME Colab, le serveur du
# modèle). Le pont porte l'URL du runtime à ton ordi, SI TU Y TIENS — la
# campagne se mène aux CINQ boutons de la cellule 6, dans le notebook même.
import pathlib
import urllib.parse
import urllib.request

CHEMIN_ENDPOINT = pathlib.Path("/content/endpoint_serving.txt")
if not CHEMIN_ENDPOINT.exists():
    raise SystemExit("PONT-ABSENT : aucun /content/endpoint_serving.txt — "
                     "la cellule 5 (le serving) d'abord.")
endpoint = CHEMIN_ENDPOINT.read_text(encoding="utf-8").strip()
port = urllib.parse.urlparse(endpoint).port
print("[pont] LA GUI 1 CLIC vit DANS LE NOTEBOOK — la cellule 6, les widgets SOUS")
print("[pont] la cellule même (aucune web app à ouvrir du navigateur). Ce qui vit")
print(f"[pont] à {endpoint} : LE SERVEUR DU MODÈLE, au runtime Colab seul.")
try:
    from google.colab import output
    url_pont = output.eval_js(f"google.colab.kernel.proxyPort({port})")
except Exception as e:
    raise SystemExit("PONT-REFUSÉ : le pont vit à la voie colab.research.google.com "
                     f"seule (l'écart nommé : {type(e).__name__} — le notebook vit-il "
                     "ouvert au site Colab même ?).")
if not url_pont:
    raise SystemExit("PONT-REFUSÉ : eval_js n'a rien rendu — le notebook vit-il ouvert "
                     "au site Colab même ?")
print(f"[pont] L'URL DU RUNTIME VUE DE TON ORDI : {url_pont}")
print(f"[pont] la santé du modèle, de ton navigateur : {url_pont}/health")
try:
    with urllib.request.urlopen(f"{endpoint}/health", timeout=10) as r:
        verdict = f"VERT ({r.status}) — {r.read(200).decode('utf-8', 'replace').strip()}"
except Exception as e:
    verdict = (f"INJOIGNABLE au runtime ({type(e).__name__}) — le serving de la cellule "
               "5 est-il vivant ? La campagne (le bouton 4) en a besoin.")
print(f"[pont] /health AU RUNTIME : {verdict}")
print("[pont] l'URL vit tant que la session vit — la même session Google ouverte ;")
print("[pont] la campagne se mène AUX CINQ BOUTONS de la cellule 6, dans le notebook.")
```
